```python
import pandas as pd 
import matplotlib.pyplot as plt 
from math import pi 
%matplotlib inline
```

```python
#Create a data frame from Messi and Ronaldo's 6 Ultimate Team data points from FIFA 18
Messi = {'Pace':95,'Shooting':98,'Passing':99,'Dribbling':99,'Defending':42,'Physical':78}
Ronaldo = {'Pace':99,'Shooting':99,'Passing':90,'Dribbling':97,'Defending':50,'Physical':92}
```

```python
data = pd.DataFrame([Messi,Ronaldo], index = ["Messi","Ronaldo"])
data
```

Plotting data in a radar is similar to plotting along a straight line , like a bar chart. We still need to provide data on where our line goes, we need to label our axes and so on. 

However, as it is a cirlce, we will also need to provide the angle at which the lines run.



Firstly, lets take a list of attributes for our lables , along with a basic count of how may there are:

```python
att = list(data)
att_No = len(att)
```

We then take a list of the value we want to plot , then copy the first value to the end. Whe we plot the data, this will be the line the radr follows:

```python
values = data.iloc[1].tolist()
values += values [:1]
values
```

So these are the point that we draw on our radar, but we will need to find the angles between each point for our lines to follow. This formula below finds these angles and assigns them to 'angles'. Then just as above, we copy the first value to the end of our array to complete the line.

```python
angles = [n / float(att_No) * 2 * pi for n in range(att_No) ]
angles += angles [:1]
```

```python
fig = plt.figure()
fig.patch.set_facecolor("#ffffff")  # clean white background
ax = plt.subplot(111, polar=True)
ax.set_facecolor("#ffffff")  # white radar area

# Axis labels
plt.xticks(angles[:-1], att, color="#000000", fontweight="bold")

# Radar plot line - vibrant electric blue
ax.plot(angles, values, color="#0066FF", linewidth=3)

# Fill - bright coral/orange
ax.fill(angles, values, color="#FF6B35", alpha=0.4)

# Title
ax.set_title("Ronaldo", fontweight="bold", color="#000000", size=16)
plt.show()
```

## Comparing two players

```python
# Find the values and angles for Messi - from the table at the top of the page
values2 = data.iloc[0].tolist()
values2 += values2[:1]

# Create the chart
fig = plt.figure(figsize=(7, 7))
fig.patch.set_facecolor("#f8f9fa")

ax = plt.subplot(111, polar=True)
ax.set_facecolor("#f8f9fa")

# Axis styling
plt.xticks(angles[:-1], att, size=10, weight='bold', color="#2d3436")
ax.grid(color="#dfe6e9", linewidth=1.2)
ax.tick_params(colors="#636e72")

# Ronaldo - vibrant orange
ax.plot(angles, values, color="#ff7675", linewidth=3, label="Ronaldo")
ax.fill(angles, values, color="#ff7675", alpha=0.25)

# Messi - electric purple
ax.plot(angles, values2, color="#6c5ce7", linewidth=3, label="Messi")
ax.fill(angles, values2, color="#6c5ce7", alpha=0.25)

# Professional title at top
ax.set_title("Messi vs Ronaldo", size=16, weight='bold', color="#2d3436", pad=20)

# Add legend
legend = ax.legend(loc='upper right', bbox_to_anchor=(1.2, 1.1), frameon=False, fontsize=11)
for text in legend.get_texts():
    text.set_weight('bold')

plt.tight_layout()
plt.show()
```
