# Kenyan Premier League 24/25 Goals For (GF) vs Goals Against (GA) 

This notebook aims to explore the kpl 24/25 season and plot goals for and against for the whole season.

## 1. Import the Required Modules

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
%matplotlib inline
```

## 2. Load the Data 

```python
table = pd.read_csv("../../datasets/kpl_table_24_25.csv")
table.head()
```

We will use Matplotlib's ".plot()". We will pass it three arguments, i.e: the data to plot along each axis and the plot type. In our case, the plot type will be 'o' to indicate that we want to plot markers.

```python
plt.plot(table['GF'], table['GA'], "o")
```

The above chart is plain and has no labels. Rather than plotting it directly, it can be created as indicated below.

```python
# Create the plot area 
fig, ax = plt.subplots()

# Set the plot size
fig.set_size_inches(7,5)

# Plot the chart like previously, but change 'o' to '*'
plt.plot(table['GF'],table['GA'], "*")

# Add labels to chart area
ax.set_title("KPL 24/25 Goals for & Against")
ax.set_xlabel("Goals For")
ax.set_ylabel("Goals Against")

# Display
plt.show()
```

Next, we will add crosshairs to display the average line. This will help to know if a point is performing well or not. 

To do this, we will use plt.plot() again, we give it 3 arguments as a minimum:
 - Type - 'k-' will give us two instructions to plot  with. K means black, - means draw a line.
 - Start/End X locations - Give these two cordinates in a list. We will calculate the average to get the coordinate.
 - Start/End Y locations - Give these two coordinates in a list. Below they are [50,20].

We will also give two optional arguments. Linestyle changes the line, in this case "." for a dotted line, lw will indicate the line width:

```python
# Create the plot area and indicate it's size
fig, ax = plt.subplots()
fig.set_size_inches(7, 5)

# Plot the chart 
plt.plot(table['GF'], table['GA'], "o")
plt.plot([table['GF'].mean(), table['GF'].mean()], [50, 20], 'k-', linestyle=':', lw=1)
plt.plot([20,50], [table['GA'].mean(), table['GA'].mean()], 'k-', linestyle=':', lw=1)

# Set title and add labels
ax.set_title("KPL 24/25 Goals for & Against")
ax.set_xlabel("Goals For")
ax.set_ylabel("Goals Against")

# Display 
plt.show()
```

In the above chart, the crosshairs show the averages and it helps us to group teams accordingly. We can also classify the quadrants with text on the chart. 

To do this we will use text(). We will give it two arguments for x and y location, and the text we want to write.:

```python
# Create the plot area and indicate it's size
fig, ax = plt.subplots()
fig.set_size_inches(7, 5)

# Plot the chart 
plt.plot(table['GF'], table['GA'], "o")
plt.plot([table['GF'].mean(), table['GF'].mean()], [50, 20], 'k-', linestyle=':', lw=1)
plt.plot([20,50], [table['GA'].mean(), table['GA'].mean()], 'k-', linestyle=':', lw=1)

# Set title and add labels
ax.set_title("KPL 24/25 Goals for & Against")
ax.set_xlabel("Goals For")
ax.set_ylabel("Goals Against")

# Add text
ax.text(19,50,"Poor attack, poor defense",color="red",size="8")
ax.text(41,20,"Strong attack, strong defense",color="red",size="8")

# Display 
plt.show()
```
