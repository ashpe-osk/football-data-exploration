# The Kenyan Primier League Table from the 24/25 season

## Table 1: Using a Simple Bar Chart 

```python
# Import the required modules
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
%matplotlib inline 
table = pd.read_csv("../../datasets/kpl_table_24_25.csv")
table.head()
```

Matplotlib allows us to use bar() function to plot our data, bar() accepts two arguments, x and height.

- x - This is the x coordinate for each bar. Spaced bars are often used in bar charts, so we need to provide a sequence we want, 'np.arange' provides this sequence easily.
- height - This indicates how high each bar will go/value of each bar. In our case, we will provide the points(P) column of the kpl table.

```python
plt.bar(x=np.arange(1,19),height=table['P'])
```

The above bar chart shows each team's point. 
However, we need to add axis labels, a title, and bar labels. 

```python
# Recreate the previous bar 
plt.bar(x = np.arange(1, 19), height = table['P'])

# Title it 
plt.title("Kenyan Premier League 24/25 Table")

# Give the x axis some labels across the tick marks 
# Argument one is the position for each label
# Argument two is the label values and the final one is to rotate our labels
plt.xticks(np.arange(1,19), table['Team'], rotation = 90)

# Title the x and y axes 
plt.xlabel("Team")
plt.ylabel("Points")

# Display the plot 
plt.show()
```

The next step is to create an array of team colours using hex codes. We will then map this array to each team.

```python
# Create an array of actual length of our bars 
# Each value is the hex code for the team's colors in order of table standings
team_colors = ["#0047AB", "#00A651", "#FFD700", 
               "#FDB913", "#DC143C", "#0066CC", 
               "#FFEB3B", "#1E3A8A", "#006B3F", 
               "#8BC34A", "#4CAF50", "#6B2C91", 
               "#FF5722", "#2E7D32", "#7B2CBF", 
               "#C62828", "#D32F2F", "#E53935"]

# To pass our array we add  colour argument to plt.bar()
plt.bar(x=np.arange(1,19), height = table['P'],color = team_colors)

# Title the bars as before 
plt.title("Kenyan Premier League 24/25 Table")
plt.xticks(np.arange(1,19), table['Team'], rotation = 90)
plt.xlabel("Team")
plt.ylabel("Points")

# Display the plot 
plt.show()
```

## Table 2: Using a Lollipop Chart 

```python
plt.hlines(y = np.arange(1,19), xmin=0, xmax=table['P'])
```

```python
plt.hlines(y=np.arange(1,19),xmin=0,xmax=table['P'],color="skyblue")
plt.plot(table['P'], np.arange(1,19), "o")
plt.yticks(np.arange(1,19), table['Team'])
plt.show()
```
